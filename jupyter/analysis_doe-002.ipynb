```python
import matplotlib.pyplot as plt
import pandas as pd
```

```python
name_doe = "doe-002"
date_trade = "20260928"
name_code = "9984"
csvfile = f"../doe/{name_doe}/{date_trade}_{name_code}.csv"
df = pd.read_csv(csvfile)
df
```

```python
cols_param = ["TYPE_MA_1", "PERIOD_MA_1"]
cols_response = ["Profit", "Transactions"]
```

```python
df[df["TYPE_MA_1"] == "PPF"]["PERIOD_MA_1"]
```

```python
plt.rcParams["font.family"] = "monospace"
plt.rcParams["font.size"] = 7

fig, axes = plt.subplots(1, 3, figsize=(9, 3.2))

name_x = "PERIOD_MA_1"
name_y = "Profit"
y_min0 = df[name_y].min()
y_max0 = df[name_y].max()
pad = (y_max0 - y_min0) * 0.05
y_min = y_min0 - pad
y_max = y_max0 + pad
list_type = ["PPF", "SMA", "EMA"]
for ax, name_type in zip(axes, list_type):
    df1 = df[df["TYPE_MA_1"] == name_type]
    #ax.scatter(df1[name_x], df1[name_y], s=3)
    ax.plot(
        df1[name_x],
        df1[name_y],
        "o-",
        markersize=4,
        linewidth=1,
    )

    ax.set_title(name_type)
    ax.set(xlabel=name_x, ylabel=name_y)
    ax.set_ylim(y_min, y_max)
    ax.grid(True)

fig.suptitle(f"Main Effect / {date_trade}_{name_code}", fontsize=9)

plt.tight_layout()
pngfile = f"../doe/{name_doe}/{date_trade}_{name_code}_main_effect_{name_y}.png"
plt.savefig(pngfile)
plt.show()
```
