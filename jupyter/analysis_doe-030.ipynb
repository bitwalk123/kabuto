# DOE-030
## MA1 / MA2 ウィンドウ確認

- 対象銘柄: 9984
- 実験計画: 完全実施要因計画 (Full Factorial Design)
- 実験パラメータ
    - PERIOD_MA_1（短周期移動平均線）
        - 30, 60, 90, 120, 150, 180, 210, 240, 270
    - PERIOD_MA_2（長周期移動平均線）
        - 300, 600, 900, 1200, 1500, 1800, 2100, 2400, 2700
- 固定パラメータ
    - TYPE_MA_1: EMA
    - START_TRAILING: 10  # トレーリング・ストップを開始する最大含み益
    - THRESHOLD_TRAILING: 0.8  # トレーリング・ストップのしきい値比
- MA1 / MA2 クロス・シグナルのエントリ・エグジット
    - クロス時に反対売買をしない。
- 評価指標
    - 収益と約定回数

```python
from pathlib import Path

import pandas as pd

# === 実験番号 ===
name_doe = "doe-030"

# pd.set_option("display.max_rows", None)
pd.set_option("display.max_columns", None)
pd.set_option("display.width", 300)

# 指定したディレクトリのPathオブジェクトを作成
dir_path = Path(f"../doe/{name_doe}")

# .glob() で .csv ファイルのリスト（イテレータ）を取得してリストに変換
list_csv_file = sorted([str(p) for p in dir_path.glob("*.csv") if p.is_file()])

# 集計対象の条件列
columns_group = [
    "TYPE_MA_1",
    "PERIOD_MA_1",
    "PERIOD_MA_2",
]

# CSV ファイルを読み込んで結合
df = pd.concat(
    [
        pd.read_csv(file).assign(
            DATE=pd.to_datetime(Path(file).stem[:8], format="%Y%m%d")
        )
        for file in list_csv_file
    ],
    ignore_index=True,
)

# 条件ごとに Profit を集計
df_total = df.groupby(columns_group, as_index=False).agg(
    Profit_Total=("Profit", "sum"),
    Profit_Median=("Profit", "median"),
    Transactions_Median=("Transactions", "median"),
)

print(df)
print("\nSummary Statistics")
print(df_total)
print("\n*** Past", len(list_csv_file), "days statistics ***")
print("\n△ Profit_Total Top 3")
print(df_total.nlargest(3, "Profit_Total"))
print("\n▼ Profit_Total Bottom 3")
print(df_total.nsmallest(3, "Profit_Total"))
print("\n\n△ Profit_Median Top 3")
print(df_total.nlargest(3, "Profit_Median"))
print("\n▼ Profit_Median Bottom 3")
print(df_total.nsmallest(3, "Profit_Median"))
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.rcParams["font.family"] = "monospace"
#plt.rcParams["font.size"] = 11

cols_param = [
    "PERIOD_MA_1",
    "PERIOD_MA_2",
]


for name_y in ["Profit", "Transactions"]:
    fig, axes = plt.subplots(1, 2, figsize=(9, 3))

    for ax, col in zip(axes, cols_param):
        sns.stripplot(
            data=df,
            x=col,
            y=name_y,
            size=3,
            jitter=0.25,  # 横方向へのばらつき具合
            alpha=0.5,  # 重なりが見やすいよう少し透明に
            color="tab:blue",  # 散布図の色を指定
            ax=ax,
        )
    
        # x の各水準ごとの平均値
        mean = df.groupby(col)[name_y].mean()
        # stripplot の x 軸位置（0, 1, 2...）に合わせて線をつなぐ
        ax.plot(
            range(len(mean)),
            mean.values,
            "o-",
            color="red",
            markersize=4,
            linewidth=1,
            alpha=0.75,
            zorder=5,  # 点の裏に隠れないよう前面に描画
        )
    
        # x の各水準ごとのメジアン
        median = df.groupby(col)[name_y].median()
        # stripplot の x 軸位置（0, 1, 2...）に合わせて線をつなぐ
        ax.plot(
            range(len(median)),
            median.values,
            "o-",
            color="orange",
            markersize=5,
            linewidth=1,
            alpha=0.75,
            zorder=5,
        )
    
        ax.set(xlabel=col, ylabel=name_y)
        ax.grid(True)
    
        #ax.set_title(f"Main Effect / {name_doe}")
    fig.suptitle(f"Main Effects / {name_doe}")
    
    plt.tight_layout()
    pngfile = f"../doe/{name_doe}/main_effect_{name_y}.png"
    plt.savefig(pngfile)
    plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Profit を Z 値とする格子状データを作成
# 同一パラメータの複数日分を統計処理
for stat in ["mean", "median"]:
    pivot = df.pivot_table(
        index="PERIOD_MA_2",
        columns="PERIOD_MA_1",
        values="Profit",
        aggfunc=stat,
    )
    
    # X, Y, Z を用意
    X, Y = np.meshgrid(
        pivot.columns,
        pivot.index,
    )
    Z = pivot.values
    
    # Contour Map
    fig, ax = plt.subplots(figsize=(6, 5))
    
    contour = ax.contourf(
        X, Y, Z,
        levels=11,
        cmap="RdYlBu_r",
    )
    
    # 等高線を重ねる（任意）
    lines = ax.contour(
        X, Y, Z,
        levels=10,
        colors="black",
        linewidths=0.5,
    )
    #ax.clabel(lines, inline=True, fontsize=8)
    ax.clabel(
        lines,
        inline=True,
        fontsize=8,  # フォントサイズを指定
        fmt="%.0f",   # 整数表示（任意）
    )
    
    # カラーバー
    fig.colorbar(contour, ax=ax, label="Profit")
    
    # グラフ領域を 1:1 の正方形にする
    ax.set_box_aspect(1)
    
    ax.set_xlabel("PERIOD_MA_1")
    ax.set_ylabel("PERIOD_MA_2")
    ax.set_title(f"Profit ({stat}) / {name_doe}")
    
    ax.grid(False)
    fig.tight_layout()
    
    pngfile = f"../doe/{name_doe}/contourmap_Profit_{stat}.png"
    plt.savefig(pngfile)
    plt.show()
```
